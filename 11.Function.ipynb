```python
num=5
fact=1
if num<0:
    print("factorial does not exist")
else:
    for i in range(1,num+1):
        fact=fact*i
        print("factorial is:",fact)
```

# 2.keyword arguments

```python
def addition(a,b,c):
    print("This is addition function")
    add=a+b+c
    print(f"Addition of number is:",add)
    return add
addition(10,20,30)
```

```python
def addition(a,b,c):
    print("This is addition function")
    add=a+b+c
    print(f"Addition of number is:",add)
    return add
addition(a=10,c=20,b=30) #we can pass value with keyword ,position not matter
```

```python
def addition(a,b,c=9):
    print("This is addition function")
    add=a+b+c
    print(f"Addition of number is:",add)
    return add
addition(a=10,b=50)  #we can change value at a calling time
```

# 4.Arbitary positional argument (*args)

```python
def avrg(*args):
    sum=0
    for i in args:
        sum+=i
        avg=sum/len(args)
    print(avg)
avrg(3,4,5,6,7)
```

# 4.Arbitary keyword argument(**kwargs)

```python
def test(a,b,c,*args,x=100,y=200,**kwargs):
    print("postional argument::",a,b,c)
    print("Arbitary positionl arguments::",args)
    print("Default argument::",x,y)
    print("Arbitary keyword arguments::",kwargs)
test(10,20,30,1,2,3,4,5,6,7,8,9,y='python',x=500,phy=98,chem=66,math=90,bio=90)
```

##### write a python program to find square and cube every number a given list of integers using lambda.

```python
l1=[1,2,3,4,5,6,1]
square=list(map(lambda x:x**2,l1))
cube=list(map(lambda x:x**3,l1))
print(square)
print(cube)
```

```python
Map()                                                       Filter()
Applies a function to each elements.                        Selects elements based on condition.
It is used to transform data.                               It is used to filter/select data.
It can change the values.                                   It keeps only the values satisfy the condition.
```
