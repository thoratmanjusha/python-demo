# copy

```python
emp_data={"name":"manjusha",
          "Age":22,
          "company name":"tcs"}
new_dict=emp_data
new_dict.update({"Salary":7999})
print(emp_data)
print(new_dict)
```

```python
emp_data={"name":"manjusha",
          "Age":22,
          "company name":"tcs"}
new_dict=emp_data.copy()
new_dict.update({"Salary":7999})
print(emp_data)
print(new_dict)
```
