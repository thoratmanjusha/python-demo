# Tuple function

## 1.index()

```python
t1=(10,20,30,40,50)
i=t1.index(40)
print(i)
```

```python
t1=(10,20,30,40,50)
i=t1.index(50)
print(i)
```

```python
t1=(10,20,30,40,50)
i=t1.index(10)
print(i)
```

```python
t1=(10,20,30,40,50)
i=t1.index(400)
print(i)
```

```python
t1=(10,20,30,40,50)
index_list=[]
for i,val in enumerate(t1):
    if val==50:
        index_list.append(i)
index_list
```

```python
t1=(10,20,30,40,50)
num=40
index_list=[i for i,val in enumerate(t1) if val==num]
index_list
```

```python
t1=(10,20,30,40,50,60,70)
num=300
index_list=tuple([i for i,val in enumerate(t1) if val==num])
index_list
```
