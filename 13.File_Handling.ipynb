# 2.readlines

```python
data="""write mode create a new text file if file does not exits.
write mode override existing file if file already exits."""
f=open("DATA.txt","w")
f.close()
print("*"*50)
print(data)
print("*"*50)
```

```python
f=open("DATA.txt","r")
data=f.read()
f.close()
print("***************************************")
data_list=data.split("\n")
print(data_list)
print("***************************************")
```

```python
f=open("DATA.txt",'r')
data=f.readlines()
f.close()
print("***************")
print(data)
print("******************")
```

# Json file

```python
import json
```

```python
emp_data={'name':'manjusha',
            'salary':400000,
            'location':'pune'}
    
```

```python
with open("emp_data.json",'w') as f:
    json.dump(emp_data,f)
```

# Read JSON files
