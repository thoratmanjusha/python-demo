# List Comprehension

```python
l1=[4,5,6,7,8]
for i in l1:
    print(i**3)
```

```python
sub_name=['Hin','Bio','Phy','Math','Chem']
new_list=[]
for sub in sub_name:
    new_list.append(sub.upper())
    print(new_list)
```

```python
l1=[2,3,3,5,6,78,89]
sq=list()
for i in l1:
    sq.append(i**2)
    print(sq)
```

```python
l1=[3,4,5,6,8]
square=[i**2 for i in l1]
square
```

```python
l1=[3,4,5,6,8]
square=[i**3 for i in l1]
square
```

```python
sub_name=['Hin','Bio','Phy','Math','Chem']
new_list=[sub.upper() for sub in sub_name]
new_list
```

```python
num=[1,2,3,4,4,5,6,7,8,8]
even_num=[]
for i in num:
    if i%2==0:
        even_num.append(i)
print(even_num)
```

```python
num=[1,2,3,4,4,5,6,7,8,8]
even_list=[i for i in num if i%2==0]
even_list
```

```python
num=[1,2,3,4,4,5,6,7,8,8]
even_list=[i for i in num if i%2!=0]
even_list
```

```python
day_list=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
result=[]
for day in day_list:
    if len(day)==6:
        result.append(day)
print(result)
```

```python
day_list=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
result=[day for day in day_list if len(day)==6]
print(result)
```
