# Nested for loop

```python
for i in range(3):
    print("hello")
    print("we are learning python")
    print("Techno savvy institute")
    print()
```

```python
for i in range(3):
    print("hello")
    print("we are learning python")
    print("Techno savvy institute")
    for j in range(4):
        print("nested for loop")
    print("*"*50)
```

```python
for char in "python":
    print(f"character is:{char}")
```

```python
for char in "python":
    print(f"character is:{char}")
    for i in range(2):
        print("Data science")
    print("-"*30)
```

```python
list1=['pune','mumbai','delhi']
for city in list1:
    print(f"city is:{city.title()}")
    for i in range(3):
        print("python and data science")
    print("-"*30)
    
```
