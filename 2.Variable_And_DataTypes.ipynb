# Types of variable

```python
#Global
a=10
b=20
c=30
def my_fun():
    print("Value of variable a is::",a)
    print("Value of variable b is::",b)
    print("Value of variable c is::",c)
my_fun()
```

```python
a=10
b=20
c=30
def my_fun():
    print("Value of variable a is::",a) #inside funct
    print("Value of variable b is::",b)
    print("Value of variable c is::",c)
my_fun()
print(a) #outside function
print(b)
print(c)
```

```python
#Local

def my_fun():
    x=10
    y=20
    z=30
    print("Value of X is::",x)
    print("Value of Y is::",y)
    print("Value of Z is::",z)
my_fun()
```

```python
def my_fun():
    x=10
    y=20
    z=30
    print("Value of X is::",x)
    print("Value of Y is::",y)
    print("Value of Z is::",z)
my_fun()
print(x)
print(y)
print(z)
```
