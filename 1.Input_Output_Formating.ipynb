# String formation

##### Type of string formating

##### 1)Formatting with string literals cell f-string

```python
a=10
b=20
add=a+b
print(f"A=={a} and B=={b} =={add}")
```

##### 2)Formating with format method

```python
a=10
b=20
add=a+b
print("Addition  id {} and {} is {}".format(a,b,add))
```

```python
a=10
b=20
add=a+b
print("Addition  id {} and {} is {}".format(b,a,add))
```

##### 3)Formatting with % operator

```python
a=10
b=20
add=a+b
print("Addition of %d and %d is : %d "%(a,b,add))
```

```python
a=10
b=20.87
add=a+b
print("Addition of %d and %f is : %f "%(a,b,add))
```

```python
a=10
b=20.98
add=a+b
print("Addition of %d and %f is : %d "%(a,b,add))
```

```python
a=100
b=200
add=a+b
print(f"Addition of A=={a} and B=={b} is :{add}")
print("Addition of {} and {} is: {}".format(a,b,add))
print("Addition of {1} and {0} is: {2}".format(a,b,add))
print("Addition of %d and %d is : %f"%(a,b,add))
print("Addition of %d and %d is : %d"%(a,b,add))
print("Addition of %d and %d is : %.2f"%(a,b,add))
print("Addition of %d and %d is : %.5f"%(a,b,add))
```
